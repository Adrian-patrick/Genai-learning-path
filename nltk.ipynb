# Demo project


```python
def clean_text(text):
    words = word_tokenize(text)
    cleaned_words = [
        word.lower() for word in words
        if word not in string.punctuation
    ]
    return cleaned_words
```

```python
def stopword_removal(words):
    stop_words = set(stopwords.words('english'))
    cleaned_words = [
        word for word in words
        if word not in stop_words
    ]
    return cleaned_words
```

```python
def lemmatizer(words):
  cleaned_words = [snowball.stem(word) for word in words]
  return cleaned_words
```

```python
def tagger(words):
  tagged_words = nltk.pos_tag(words)
  return tagged_words

```

```python
def text_cleaner(text):
  cleaned_words = clean_text(text)
  print("words lowered punctuation removed and tokenized ->",cleaned_words)
  cleaned_words = stopword_removal(cleaned_words)
  print("\nstopwords removed ->",cleaned_words)
  cleaned_words = lemmatizer(cleaned_words)
  print("\nwords lemmatized ->",cleaned_words)
  cleaned_words = tagger(cleaned_words)
  print("\nwords pos tagged ->",cleaned_words)
  return
```

```python
text_cleaner(raw_string)
```
